### 1. Remove Unwanted HTML Tags

````python
from bs4 import BeautifulSoup

def strip_html_noise(html_content):
    soup = BeautifulSoup(html_content, "html.parser")
    
    # Define tags that do not contain core textual knowledge
    unwanted_tags = [
        "script", "style", "noscript", "header", "footer", 
        "nav", "aside", "form", "button", "svg", "iframe"
    ]
    
    for tag in soup(unwanted_tags):
        tag.decompose()  # Completely removes the element from the parse tree
        
    return soup
````

### 2. Extract Text and Normalize Whitespace

````python
import re

def extract_and_normalize(soup):
    # Extract plain text with a separator to keep layout boundaries
    # Pass strip=True to remove outer padding from HTML block elements before joining them
    # eg: removes "\n\n\n\n\n....."
    raw_text = soup.get_text(separator=" ", strip=True)

    # 1. Replace multiple spaces/tabs with a single space
    cleaned = re.sub(r'[ \t]+', ' ', raw_text)
    
    # 2. Collapse three or more consecutive newlines down to just two
    cleaned = re.sub(r'\n{3,}', '\n\n', cleaned)
    
    # 3. Strip out blank spaces leading or trailing the text block
    cleaned = cleaned.strip()
    
    # 4. Remove excessive newlines
    cleaned = re.sub(r'\n+', '\n', cleaned)

    return cleaned
````

### 3. Filter Boilerplate and Low-Quality Text

````python
def is_high_quality(text):
    # Filter out sentences that are too short (likely menu buttons or dates)
    if len(text.split()) < 5:
        return False
        
    # Filter out common legal/web boilerplate patterns
    boilerplate_patterns = [
        r"copyright ©", r"all rights reserved", r"cookie policy", 
        r"terms of service", r"privacy policy", r"sign up for our newsletter"
    ]
    
    for pattern in boilerplate_patterns:
        if re.search(pattern, text, re.IGNORECASE):
            return False
            
    return True

def filter_paragraphs(cleaned_text):
    # Split text into paragraphs and keep only high-quality ones
    paragraphs = cleaned_text.split('\n\n')
    filtered_paragraphs = [p for p in paragraphs if is_high_quality(p)]
    
    return "\n\n".join(filtered_paragraphs)
````

### 4. Deduplicate and Standardize Text

````python
import unicodedata

def final_polish(text):
    # Normalize Unicode characters (e.g., converts '\xa0' to normal spaces)
    text = unicodedata.normalize("NFKC", text)
    
    # Remove bracketed web references like [1], [2], [citation needed]
    text = re.sub(r'\[\d+\]', '', text)
    text = re.sub(r'\[citation needed\]', '', text, flags=re.IGNORECASE)

    # Option A: Collapse everything into a single space (Best for continuous text streams)
    # clean_text = re.sub(r'\s{2,}', ' ', text)
    
    # Option B: Collapse multiple newlines into a single standard line break
    # clean_text = re.sub(r'\n{2,}', '\n', text)
    
    return text
````

### 5. Document Aggregation and Boundary Marking
During pre-training, the model needs to know where one webpage ends and a completely unrelated one begins.

If you just concatenate all text blindly, the model will try to learn relationships between unrelated websites.

Use a special end-of-text token—such as `<|endoftext|>` for GPT-style models 

or `</s>` for Llama-style models—to separate your scraped documents.

````python
def format_for_pretraining(cleaned_documents, eos_token="<|endoftext|>"):
    formatted_dataset = []
    
    for doc in cleaned_documents:
        # Append the special token to mark the end of a cohesive document
        formatted_text = f"{doc} {eos_token}"
        formatted_dataset.append({"text": formatted_text})
        
    return formatted_dataset
````

For the **Qwen3.5-9B** family, the standard end-of-text control token used for unsupervised pre-training and document separation is `<|endoftext|>`.

While Qwen's instruction/chat models rely on `<|im_end|>` to mark the end of an individual speaker's turn, the base model architecture utilizes `<|endoftext|>` as the explicit **end-of-document boundary** indicator to manage packed text streams.

### Critical Warning: The Qwen Tokenizer Pitfall
When setting up your dataset for pre-training, there is a known architectural hazard with the Qwen tokenizer: 
**By default, Qwen maps both its `eos_token` and pad_token to `<|endoftext|>`**.

During pre-training, standard training loops mask out pad_token losses. Because the padding token is identical to the end-of-text token, the model can struggle to learn when a document actually ends. This frequently causes the model to hallucinate and endlessly loop random text during inference.

### Correct Code Configuration for Qwen3.5-9B
To safely inject the boundary markers and fix the padding overlap, configure your tokenization and dataset setup like this:
```python
from transformers import AutoTokenizer

model_id = "Qwen/Qwen3.5-9B-Base"
tokenizer = AutoTokenizer.from_pretrained(model_id)

# 1. Verify your end-of-text token string
# This outputs: "<|endoftext|>"
eos_string = tokenizer.eos_token  

# 2. FIX THE PADDING OVERLAP
# Use the dedicated '<|vision_pad|>' or an unused control token for padding
# This ensures '<|endoftext|>' loss is computed correctly during training
if "<|vision_pad|>" in tokenizer.get_vocab():
    tokenizer.pad_token = "<|vision_pad|>"
else:
    tokenizer.pad_token = "<|extra_0|>" # Safe fallback for Qwen

```

### 6. Pack Data into Fixed Block Sizes (Packing)
LLMs train on fixed context window lengths (e.g., 2048, 4096, or 8192 tokens). 

If your scraped webpages are short, training on one page per row creates massive padding noise, which wastes compute.

"Packing" joins all your text together into one giant stream and cuts it into exact chunks matching your model's maximum sequence length.

````python
from transformers import AutoTokenizer

def pack_dataset(dataset, model_id, max_length=2048):
    tokenizer = AutoTokenizer.from_pretrained(model_id)
    
    # 1. Tokenize the entire dataset text into one long list of token IDs
    all_token_ids = []
    for row in dataset:
        tokens = tokenizer.encode(row["text"], add_special_tokens=False)
        all_token_ids.extend(tokens)
        
    # 2. Slice the token stream into chunks of exactly max_length
    packed_chunks = []
    for i in range(0, len(all_token_ids), max_length):
        chunk = all_token_ids[i : i + max_length]
        
        # Keep only full-sized chunks to maximize compute efficiency
        if len(chunk) == max_length:
            packed_chunks.append({
                "input_ids": chunk,
                "labels": chunk.copy()  # For causal LM, labels are identical to inputs
            })
            
    return packed_chunks
````

The data packing step is actually **fully included and handled automatically** inside the fine-tuning code. It was not left out; it was simply offloaded to the Hugging Face trl library to save you from writing complex manual loop code.

In the provided script, the packing step is executed by these two specific lines inside the **`SFTTrainer`**:

```python
trainer = SFTTrainer(
    ...
    max_seq_length=4096,        # <- Dictates the fixed block size
    packing=True,               # <- ENABLES STEP 6 (DATA PACKING)
    ...
)
```
#### How the Code Executes Step 6 Under the Hood
When you supply `packing=True` and a `max_seq_length` to the `SFTTrainer`, the library internally executes your entire original Step 6 logic:
1. **Automatic Tokenisation**: It reads the clean strings from your `dataset`.
2. **Boundary Appending**: It automatically attaches the Qwen end-of-text token (`<|endoftext|>`) to the end of every individual document.
3. **Stream Concatenation**: It joins all those tokens into a single, continuous, massive sequence.
4. **Fixed-Size Chunking**: It cuts that endless stream of tokens into precise blocks of `exactly max_seq_length=4096` tokens.
#### Why It Is Better to Let `SFTTrainer` Do It
- **No Wasted Memory**: If you pack tokens manually into a JSON dataset *before* training, saving it creates large files. `SFTTrainer` handles this dynamically in your system RAM right as training begins.
- **Guaranteed Zero Padding**: The trainer ensures that every single block fed to your GPU has exactly 4,096 tokens. There are no trailing zeros, meaning you do not waste GPU compute cycles on padding calculations.

The cleaned web-scraped data goes straight into that script, and trl handles the formatting seamlessly.

````python
import re
from datasets import Dataset
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments
from peft import LoraConfig, get_peft_model
from trl import SFTTrainer

# ==========================================
# 1. LOAD YOUR CLEANED BS4 DATA
# ==========================================
# scraped_documents = [
#     "Web page text content here with gaps \n \n \n \n that get removed.",
#     "Another completely independent web page text line..."
# ]

# def clean_gaps(text):
#     # Strip multi-newline gaps and extra whitespaces
#     return re.sub(r'\s{2,}', ' ', text).strip()

# cleaned_data = [{"text": clean_gaps(doc)} for doc in scraped_documents]
cleaned_data = [{"text": doc} for doc in scraped_data]
dataset = Dataset.from_list(cleaned_data)
dataset
````

````python
# ==========================================
# 2. CONFIGURATOR TOKENIZER & MODEL
# ==========================================
model_id = "Qwen/Qwen3.5-9B-Base"
tokenizer = AutoTokenizer.from_pretrained(model_id)

# Fix Qwen padding conflict with <|endoftext|>
if "<|vision_pad|>" in tokenizer.get_vocab():
    tokenizer.pad_token = "<|vision_pad|>"
else:
    tokenizer.pad_token = "<|extra_0|>"

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    device_map="auto",
    torch_dtype="bfloat16" # Essential for Qwen 3.5 architecture
)
````

````python
# ==========================================
# 3. DEFINE LORA CONFIGURATION
# ==========================================
peft_config = LoraConfig(
    r=16,                       # Rank: higher means more capacity, 16 or 32 is optimal for pre-training
    lora_alpha=32,              # Scaling factor (usually 2x Rank)
    target_modules=[            # Target all linear layers for comprehensive domain learning
        "q_proj", "k_proj", "v_proj", "o_proj", 
        "gate_proj", "up_proj", "down_proj"
    ],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

# Apply LoRA parameters to the base model
model = get_peft_model(model, peft_config)
model.print_trainable_parameters()  # Displays the percentage of parameters being trained
````

````python
# ==========================================
# 4. DEFINE TRAINING ARGUMENTS
# ==========================================
training_args = TrainingArguments(
    output_dir="./qwen3.5_lora_finetuning",
    learning_rate=2e-4,         # LoRA requires a slightly higher learning rate than full fine-tuning
    per_device_train_batch_size=2,
    gradient_accumulation_steps=8,
    num_train_epochs=1,
    weight_decay=0.01,
    warmup_ratio=0.03,
    logging_steps=10,
    save_strategy="steps",
    save_steps=500,
    bf16=True,                  # Set to False and fp16=True if your GPU is older than Ampere (e.g., V100)
    fp16=False,
)
````

````python
# ==========================================
# 5. INITIALIZE TRL SFTTRAINER WITH PACKING
# ==========================================
trainer = SFTTrainer(
    model=model,
    train_dataset=dataset,
    dataset_text_field="text",
    max_seq_length=4096,        # Context window packing size
    packing=True,               # Packs raw texts without padding
    tokenizer=tokenizer,
    args=training_args,
)
````

````python
# ==========================================
# 6. START PRE-TRAINING
# ==========================================
trainer.train()

# Save the adapter weights separately at the end
trainer.model.save_pretrained("./qwen3.5_lora_adapter")
````
